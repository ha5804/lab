전체 결론

대부분 데이터셋에서 공통적으로:

P-AUROC > I-AUROC

즉 AdaptCLIP은 이미지 전체가 이상인지 판단하는 것보다, pixel/patch 단위로 이상 점수를 주는 능력이 더 강하게 나오는 경향이 있음.

하지만 동시에:

P-AUROC는 높은데 P-AUPR은 낮은 클래스가 많음

즉 대충 이상 위치 ranking은 잘하지만, 실제 mask를 정밀하게 찍는 능력은 약한 경우가 많다고 볼 수 있음.

1. MVTec

평균:

I-AUROC 93.5
P-AUROC 90.9

여기는 특이하게 I-AUROC가 평균적으로 더 높음.

P-AUROC가 더 높은 클래스
hazelnut, capsule, pill, screw, toothbrush, bottle

특히:

screw: I 85.9 → P 97.9

screw는 이미지 전체로는 애매하지만, local defect는 잘 잡는 케이스.

I-AUROC가 더 높은 클래스
carpet, leather, cable, transistor, metal_nut, zipper, tile, wood, grid

특히:

transistor, metal_nut, cable

은 pixel localization이 약함.

P-AUROC와 P-AUPR gap 큰 클래스

강한 예외:

leather, grid, screw, toothbrush, cable, capsule

특히:

leather: P-AUROC 99.1 / P-AUPR 27.9
grid: 97.4 / 27.5
screw: 97.9 / 28.8

해석:

위치 순위는 잘 잡지만, anomaly mask가 퍼지거나 noisy할 가능성이 큼.

거의 없음. 사실상:
